# Notebook 01 — Project Setup and Environment

## Objective
Verify that the IPL Analytics V2 project uses the intended Python
environment, required dependencies, and project directory structure.

## Acceptance criteria
- Python 3.12 is active.
- Required packages import successfully.
- The project root is detected correctly.
- Required directories exist.
- No credentials are exposed in notebook output.

```python
import sys
import platform
import importlib.metadata as metadata
from pathlib import Path

print("Python version:", sys.version)
print("Python executable:", sys.executable)
print("Operating system:", platform.platform())
print("Working directory:", Path.cwd())
```

```python

from pathlib import Path

current_dir = Path.cwd().resolve()

# Locate the project root by finding its expected folders.
if (current_dir / "notebooks").is_dir() and (current_dir / "src").is_dir():
    PROJECT_ROOT = current_dir

elif (
    current_dir.name.lower() == "notebooks"
    and (current_dir.parent / "src").is_dir()
):
    PROJECT_ROOT = current_dir.parent

elif (
    (current_dir.parent / "notebooks").is_dir()
    and (current_dir.parent / "src").is_dir()
):
    PROJECT_ROOT = current_dir.parent

else:
    raise FileNotFoundError(
        "Could not locate the IPL Analytics V2 project root. "
        "Open this notebook from the project's notebooks directory."
    )

print("Detected project root:", PROJECT_ROOT)
print("Project root exists:", PROJECT_ROOT.is_dir())
print("Notebooks directory exists:", (PROJECT_ROOT / "notebooks").is_dir())
print("Source directory exists:", (PROJECT_ROOT / "src").is_dir())

assert PROJECT_ROOT.is_dir()
assert (PROJECT_ROOT / "notebooks").is_dir()
assert (PROJECT_ROOT / "src").is_dir()

print("\nPROJECT ROOT DETECTION: PASS")
```

```python
import importlib

required_packages = [
    "numpy",
    "pandas",
    "matplotlib",
    "seaborn",
    "sklearn",
    "scipy",
    "yaml",
    "dotenv",
    "joblib",
    "pytest",
    "IPython",
]

failed_imports = []

for package in required_packages:
    try:
        importlib.import_module(package)
        print(f"PASS: {package}")
    except Exception as exc:
        failed_imports.append((package, str(exc)))
        print(f"FAIL: {package} — {exc}")

assert not failed_imports, (
    f"Dependency verification failed: {failed_imports}"
)

print("\nAll required imports passed.")
```

```python
distribution_names = [
    "numpy",
    "pandas",
    "matplotlib",
    "seaborn",
    "scikit-learn",
    "scipy",
    "PyYAML",
    "python-dotenv",
    "joblib",
    "pytest",
    "ipykernel",
    "jupyter",
]

versions = {}

for package in distribution_names:
    try:
        versions[package] = metadata.version(package)
    except metadata.PackageNotFoundError:
        versions[package] = "NOT INSTALLED"

for package, version in versions.items():
    print(f"{package}: {version}")
```

```python
required_directories = [
    "config",
    "data/raw",
    "data/interim",
    "data/processed",
    "data/external",
    "data/samples",
    "notebooks",
    "src/ipl_analytics/data",
    "src/ipl_analytics/features",
    "src/ipl_analytics/models",
    "src/ipl_analytics/database",
    "models/batting",
    "models/bowling_wickets",
    "models/bowling_runs_conceded",
    "models/manifests",
    "artifacts/metrics",
    "artifacts/reports",
    "artifacts/figures",
    "tests",
    "scripts",
    "docs",
    "deployment",
]

missing_directories = [
    directory
    for directory in required_directories
    if not (PROJECT_ROOT / directory).is_dir()
]

for directory in required_directories:
    status = "PASS" if directory not in missing_directories else "FAIL"
    print(f"{status}: {directory}")

assert not missing_directories, (
    f"Missing directories: {missing_directories}"
)

print("\nAll required directories exist.")
```
